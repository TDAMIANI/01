Como dijimos en clases anteriores, Python tiene implementadas muchas librerias para poder trabajar con datos. En la clase de hoy trabajaremos con una de ellas: `Numpy`.

Antes de comenzar, vamos a hablar un poco de esta libreria o modulo.

**Numpy** es una librería optimizada para realizar cálculos numéricos con vectores y matrices. A diferencia de otros lenguajes de programación, Python no posee en su estructura central la figura de matrices. Eso quiere decir que para poder trabajar con esta estructura de datos deberiamos trabajar con listas de listas. NumPy introduce el concepto de arrays o matrices.

Al ser de código abierto, `numpy` posee una documentación muy amplia que es **SIEMPRE RECOMENDABLE** consultar.

- [Documentacion NumPy](https://devdocs.io/numpy/)

# Ejercicios

1. Crear un arreglo de ceros de longitud 12
2. Crear un arreglo de longitud 10 con ceros en todas sus posiciones y un 10 en la posición número 5
3. Crear un arreglo que tenga los números del 10 al 49
4. Crear una arreglo 2d de shape (3, 3) que tenga los números del 0 al 8
5. Crear un arreglo de números aleatorios de longitud 100 y obtener su media y varianza
6. Calcular la media de un arreglo usando np.sum
7. Calcular la varianza de un arreglo usando np.sum y np.mean
8. Crear un array de números aleatorios usando np.random.randn.


```python
import numpy as np
```

```python
# 1. Crear un arreglo de ceros de longitud 12
array1 = np.zeros(12)
```

```python
print(array1)
```

```python
##Crear un arreglo de longitud 10 con ceros en todas sus posiciones y un 10 en la posición número 5
array2 = np.zeros(10)
array2[4] = 10
print(array2)
```

```python
##Crear un arreglo que tenga los números del 10 al 49
array3 = np.arange(10, 50)
print(array3)
```

```python
# 4. Crear una arreglo 2D de shape (3, 3) que tenga los números del 0 al 8
array4 = np.arange(9).reshape(3, 3)
print(array4)
```

```python
# 5. Crear un arreglo de números aleatorios de longitud 100 y obtener su media y varianza
random_array = np.random.rand(100)
mean = np.mean(random_array)
variance = np.var(random_array)
print(mean)
print(variance)
```

```python
# 6. Calcular la media de un arreglo usando np.sum
array_mean = np.array([1, 2, 3, 4, 5])
mean_calculated = np.sum(array_mean) / array_mean.size
print(mean_calculated)
```

```python
## Calcular la varianza de un arreglo usando np.sum y np.mean
array_variance = np.array([1, 2, 3, 4, 5])
mean_array = np.mean(array_variance)
variance_calculated = np.sum((array_variance - mean_array) ** 2) / array_variance.size
print(array_variance)
print(mean_array)
print(variance_calculated)
```

```python
# 8. Crear un array de números aleatorios usando np.random.randn
random_array_np = np.random.randn(10)
print(random_array_np)
```

```python

# Imprimir los resultados
print("1. Arreglo de ceros:", array1)
print("2. Arreglo con un 10 en la posición 5:", array2)
print("3. Arreglo con números del 10 al 49:", array3)
print("4. Arreglo 2D del 0 al 8:", array4)
print("5. Media del arreglo de números aleatorios:", mean)
print("   Varianza del arreglo de números aleatorios:", variance)
print("6. Media del arreglo usando np.sum:", mean_calculated)
print("7. Varianza del arreglo usando np.sum y np.mean:", variance_calculated)
print("8. Arreglo de números aleatorios con np.random.randn:", random_array_np)





```
